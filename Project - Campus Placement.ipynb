NUMPY ANALYSIS

```python
aptitude = np.array(df["Aptitude_Score"])
```

```python
print("Minimum Aptitude Score =", np.min(aptitude))
print("Maximum Aptitude Score =", np.max(aptitude))
print("Total Aptitude Score =", np.sum(aptitude))
print("Average Aptitude Score =", np.mean(aptitude))
```

```python
high_scores = aptitude[aptitude > 80]
print(high_scores)
```

```python
print("Number of students scoring above 80 =", len(high_scores))
```

```python
#placement ready
def placement_readiness(score):
    if score >= 80:
        return "High"
    elif score >= 60:
        return "Moderate"
    else:
        return "Needs Improvement"
```

```python
df["Overall_Score"] = (
    df["Assignment_Score"] +
    df["Project_Score"] +
    df["Aptitude_Score"] +
    df["Communication_Score"]
) / 4
```

```python
df["Readiness"] = df["Overall_Score"].apply(placement_readiness)
df[["Student_ID", "Overall_Score", "Readiness"]].head(10)
```

```python
readiness_count = df["Readiness"].value_counts()
readiness_count
```

Course-wise Attendance vs Project Score

```python
# READINESS WISE PLACEMENT

readiness_count = df["Readiness"].value_counts()

readiness_placement = pd.crosstab(
    df["Readiness"],
    df["Placement"]
)

readiness_placement
```
